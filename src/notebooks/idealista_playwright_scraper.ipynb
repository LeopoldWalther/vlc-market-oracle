# Idealista-Einzellisting mit Playwright


```python
import asyncio
from patchright.async_api import async_playwright
import nest_asyncio

# Erlaubt die Nutzung von asyncio.run() in bestehenden Event-Loops (Jupyter/IPython)
nest_asyncio.apply()

async def main():
    async with async_playwright() as p:
        # Browser sichtbar starten
        browser = await p.chromium.launch(headless=False)

        # Kontext mit spanischer Spracheinstellung
        context = await browser.new_context(
            locale="es-ES",
            viewport={"width": 1920, "height": 1080}
        )

        page = await context.new_page()

        print("Aufruf der URL...")
        response = await page.goto(
            "https://www.idealista.com/inmueble/106749418/",
            wait_until="domcontentloaded"
        )

        print(f"HTTP-Status: {response.status}")
        print(f"Seitentitel: {await page.title()}")

        # 5 Sekunden warten zur Kontrolle
        await page.wait_for_timeout(5000)

        await context.close()
        await browser.close()

# Startet die Coroutine sicher
asyncio.run(main())
```
